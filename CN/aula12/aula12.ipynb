```python
import numpy as np

def resol_sist_triang(a, b):
    """Resolve um sistema de equações lineares Ax = b
    Parâmetros:
    a : array_like
        Matriz dos coeficientes (triangular superior).
    b : array_like
        Vetor dos termos independentes.
    """
    n = len(b)
    x = np.zeros(len(b))
    x[n-1] = b[n-1] / a[n-1,n-1]

    for k in range(n-2, -1, -1):
        s = 0
        for j in range(k+1, n):
            s += a[k,j] * x[j]
        x[k] = (b[k] - s) / a[k,k]

    return x
```

```python
#ex01 LRU
import numpy as np

A0 = np.array([[3, 2, 4],
               [1, 1, 2],
               [4, 3, -2]], dtype=float)
print("A0:\n", A0)

# Multiplicadores da primeira coluna
m21 = A0[1,0] / A0[0,0]
m31 = A0[2,0] / A0[0,0]

M0 = np.array([[1, 0, 0],
               [-m21, 1, 0],
               [-m31, 0, 1]], dtype=float)

print("M0:\n", M0)

B = np.array([[1],
              [2],
              [3]], dtype=float)

C = M0 @ B
print("C:\n", C)

A1 = M0 @ A0
print("A1:\n", A1)
# print("A1:\n", A1)

# Multiplicador da segunda coluna (linha 3, coluna 2)
m32 = A1[2,1] / A1[1,1]

M1 = np.array([[1, 0, 0],
               [0, 1, 0],
               [0, -m32, 1]], dtype=float)

D = M1 @ C
print("D:\n", D)

A2 = M1 @ A1
# print("A2:\n", A2)

L = np.array([[1, 0, 0],
              [m21, 1, 0],
              [m31, m32, 1]], dtype=float)
# print("L:\n", L)

L @ A2  # Deve resultar em A0
print("L @ A2:\n", L @ A2)
print("A2:\n", A2)

x = resol_sist_triang(A2, D.flatten())
print("Sol:", x)
```
