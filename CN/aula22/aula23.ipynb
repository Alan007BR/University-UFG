```python
import math

t = [0, -math.sqrt(3/5), math.sqrt(3/5)]
w = [8/9, 5/9, 5/9]

def gauss3():
    soma = 0
    for ti, wi in zip(t, w):
        x = 0.5*ti + 0.5
        soma += wi * math.exp(x)
    return 0.5 * soma

print(gauss3())
```
