## Resolução de sistemas triangulares

usando o método da substituição regressiva

```python
import numpy as np

# Método da substituição regressiva
def resol_sist_triang(a, b):
    """Resolve um sistema de equações lineares Ax = b
    Parâmetros:
    a : array_like
        Matriz dos coeficientes (triangular superior).
    b : array_like
        Vetor dos termos independentes.
    """
    n = len(b)
    x = np.zeros(len(b))
    x[n-1] = b[n-1] / a[n-1,n-1]

    for k in range(n-2, -1, -1):
        s = 0
        for j in range(k+1, n):
            s += a[k,j] * x[j]
        x[k] = (b[k] - s) / a[k,k]

    return x

#Exemplo de uso
A = np.array([[3, 2, 4],
              [0, 1/3, 2/3],
              [0, 0, 8]], dtype=float)

b = np.array([1, 5/3, 0], dtype=float)
x = resol_sist_triang(A, b)
print("Sol:", x)
```

```python
import numpy as np
def resol_sist_triang_inf(a, b):
    """
    Resolve um sistema de equações lineares Ax = b,
    onde A é uma matriz triangular inferior.
    Parâmetros:
    a : array_like
        Matriz dos coeficientes (triangular inferior).
    b : array_like
        Vetor dos termos independentes.
    """
    n = len(b)
    x = np.zeros(n)
    x[0] = b[0] / a[0, 0]
    for k in range(1, n):
        s = 0
        for j in range(k):
            s += a[k, j] * x[j]
        x[k] = (b[k] - s) / a[k, k]
    return x

# Exemplo de uso
A = np.array([[1, 0, 0],
              [2, 1, 0],
              [2, -3, 1]], dtype=float)

b = np.array([-3, -1, -23], dtype=float)
x = resol_sist_triang_inf(A, b)
print("Sol:", x)
```
