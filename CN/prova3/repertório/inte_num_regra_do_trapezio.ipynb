```python
import math

def trapezios_composta(f, a, b, n, mostrar_passos=True):
    """
    Aproxima ∫_a^b f(x) dx pela Regra dos Trapézios composta com n subintervalos.

    f : função Python (ex: math.exp)
    a, b : limites de integração
    n : número de subintervalos
    mostrar_passos : se true imprime os passos, tentei deixar igual os slides

    Retorna: (aprox, h, xs, fs)
    """
    h = (b - a) / n
    xs = [a + i*h for i in range(n + 1)]
    fs = [f(x) for x in xs]

    soma_interna = sum(fs[1:-1])
    aprox = (h/2) * (fs[0] + 2*soma_interna + fs[-1])

    if mostrar_passos:
        print(f"=== Regra dos Trapézios: n = {n} subintervalos ===")
        print(f"a = {a}, b = {b}, h = (b - a)/n = {h}")
        print("Pontos x_i:")
        for i, x in enumerate(xs):
            print(f"  x_{i} = {x}")
        print("\nValores f(x_i):")
        for i, (x, fx) in enumerate(zip(xs, fs)):
            print(f"  f(x_{i}) = f({x}) = {fx}")
        print("\nFórmula:")
        print("∫_a^b f(x) dx ≈ h/2 * [ f(x0) + 2*(f(x1)+...+f(x{n-1})) + f(xn) ]")
        print(f"= {h}/2 * [ {fs[0]} + 2*({soma_interna}) + {fs[-1]} ]")
        print(f"= {aprox}\n")

    return aprox, h, xs, fs
```

```python
def f_exp(x):
    return math.exp(x)

a, b = 0.0, 1.0
valor_exato = math.e - 1  # integral de exp(x) de 0 a 1

for n in (2, 5, 10):
    aprox, h, xs, fs = trapezios_composta(f_exp, a, b, n, mostrar_passos=True)
    erro = abs(valor_exato - aprox)
    print(f"n = {n}: Aproximação = {aprox:.9f}, Exato ≈ {valor_exato:.9f}, Erro ≈ {erro:.2e}\n")
```

```python
import math
from typing import Callable, Tuple, List

def trapezios_composta_latex(
    f: Callable[[float], float],
    a: float,
    b: float,
    n: int,
    func_name: str = r"f"
) -> Tuple[float, float, List[float], List[float], str]:
    """
    Regra dos Trapézios composta + string LaTeX para usar com display(Math(...)).

    f         : função Python (ex: math.exp)
    a, b      : limites de integração
    n         : número de subintervalos
    func_name : texto que aparece no integrando (ex: 'e^x', 'f')

    Retorna: (aprox, h, xs, fs, latex_str)
    """
    # malha
    h = (b - a) / n
    xs = [a + i*h for i in range(n + 1)]
    fs = [f(x) for x in xs]

    soma_interna = sum(fs[1:-1])
    aprox = (h/2) * (fs[0] + 2*soma_interna + fs[-1])

    # ----- monta pedaço simbólico: f(x0) + 2f(x1) + ... + f(xn)
    sym_terms = []
    sym_terms.append(fr"{func_name}\big({xs[0]:.3g}\big)")
    for i in range(1, n):
        sym_terms.append(fr"2{func_name}\big({xs[i]:.3g}\big)")
    sym_terms.append(fr"{func_name}\big({xs[-1]:.3g}\big)")
    symbolic = " + ".join(sym_terms)

    # ----- monta pedaço numérico: f(x0) + 2·f(x1) + ... + f(xn)
    num_terms = []
    num_terms.append(f"{fs[0]:.6g}")
    for i in range(1, n):
        num_terms.append(fr"2\cdot {fs[i]:.6g}")
    num_terms.append(f"{fs[-1]:.6g}")
    numeric = " + ".join(num_terms)

    # strings já formatadas para não ter confusão com chaves
    a_s = f"{a:.3g}"
    b_s = f"{b:.3g}"
    h_s = f"{h:.4g}"
    aprox_s = f"{aprox:.9f}"

    # LaTeX final
    latex_str = (
        rf"\int_{{{a_s}}}^{{{b_s}}} {func_name}\,dx"
        rf"\approx \frac{{{h_s}}}{{2}}\left[{symbolic}\right]"
        rf" = \frac{{{h_s}}}{{2}}\left[{numeric}\right]"
        rf" = {aprox_s}"
    )

    return aprox, h, xs, fs, latex_str
```

```python
from IPython.display import Math, display
import math

def f_exp(x):
    return math.exp(-(x/2)**2)

a, b = 0.0, 2.0

# Usa a regra dos trapézios composta para vários n
for n in (2, 5, 10, 20):
    aprox, h, xs, fs, latex_str = trapezios_composta_latex(
        f_exp, a, b, n,
        func_name=r"e^{-(x/2)^2}"
    )

    print(f"Intervalo [{a}, {b}] com n={n} subintervalos")
    print(f"Aproximação = {aprox:.9f}")
    print(f"Erro ≈ {abs(aprox - valor_exato):.2e}\n")

    # Renderiza expressão LaTeX mostrando a fórmula aplicada
    display(Math(latex_str))
```
