```python
from IPython.display import Math, display
import math

# Gauss–Legendre com 3 pontos em [a, b] para f(x) = 1/2 e^{-|x|}
# Formatação detalhada igual aos outros exemplos

def f_laplace(x):
    return math.exp(-(x/2)**2)

# Nós e pesos de Gauss–Legendre em [-1, 1] para n=3
# t_k: [-sqrt(3/5), 0, +sqrt(3/5)] ; w_k: [5/9, 8/9, 5/9]
SQ35 = math.sqrt(3.0/5.0)
GL_NODES = [-SQ35, 0.0, SQ35]
GL_WEIGHTS = [5.0/9.0, 8.0/9.0, 5.0/9.0]

# Intervalo alvo
A, B = -0.5, 0.5

# Mudança de variável: x = (a+b)/2 + t_k * (b-a)/2
mid = 0.5 * (A + B)
half = 0.5 * (B - A)

# Valor exato para comparação
valor_exato = 1.0 - math.exp(-0.5)

print("=== Gauss–Legendre (n = 3 pontos) para f(x) = 1/2 e^{-|x|} em [-0.5, 0.5] ===\n")
print("Mudança de variável:")
print("x_k = (a+b)/2 + t_k (b-a)/2")
print(f"(a+b)/2 = {mid}")
print(f"(b-a)/2 = {half}\n")

# Construção dos x_k e y_k = f(x_k) com tabela
rows = []
for k, (tk, wk) in enumerate(zip(GL_NODES, GL_WEIGHTS)):
    xk = mid + tk * half
    yk = f_laplace(xk)
    rows.append((k, tk, xk, wk, yk))

print("k    t_k           x_k                  w_k         y_k = f(x_k)")
for k, tk, xk, wk, yk in rows:
    print(f"{k:<1}    {tk:+.9f}   {xk: .9f}   {wk: .9f}   {yk: .9f}")

# Fórmula: I ≈ (b-a)/2 * sum_{k=0}^{2} w_k f(x_k)
weighted_sum = sum(wk * f_laplace(mid + tk * half) for tk, wk in zip(GL_NODES, GL_WEIGHTS))
I_gl = half * weighted_sum

print("\nFórmula:")
print("I ≈ (b-a)/2 * [ w0 f(x0) + w1 f(x1) + w2 f(x2) ]")
num_terms = " + ".join(
    [f"{wk:.9f}·{f_laplace(mid + tk*half):.9f}" for tk, wk in zip(GL_NODES, GL_WEIGHTS)]
)
print(f"= {half:.9f} * [ {num_terms} ]")
print(f"= {I_gl:.9f}\n")

print(f"Valor exato = {valor_exato:.9f}")
print(f"Erro ≈ {abs(I_gl - valor_exato):.2e}\n")

# LaTeX compacto da montagem
latex_compacto = (
    rf"I \approx \frac{{b-a}}{{2}}\sum_{{k=0}}^2 w_k f\big(x_k\big)"
    + rf" = \frac{{{half:.6f}}}{{1}}\Big["
    + " + ".join([rf"{wk:.6f}\cdot {f_laplace(mid + tk*half):.6f}" for tk, wk in zip(GL_NODES, GL_WEIGHTS)])
    + rf"\Big] = {I_gl:.9f}"
)

display(Math(latex_compacto))
```
