```python
import numpy as np
n = 6
x = np.linspace(0.0, 1.0, n)

h = x[1] - x[0]

f = np.exp(x)

I_trap = h/2 * (f[0] + 2.0 * np.sum(f[1:-1]) + f[-1])

print(I_trap)
```

```python
m = 6 # subdivisões (deve ser par)
n = m + 1
x = np.linspace(0.0, 1.0, n)
h = x[1] - x[0]
f = np.exp(x)

odd_sum = np.sum(f[1:-1:2])   # f(x1), f(x3), f(x5)
even_sum = np.sum(f[2:-1:2])  # f(x2), f(x4)
I_simpson = h/3.0 * (f[0] + f[-1] + 4*odd_sum + 2*even_sum)
print(I_simpson)
```
