# Conversão binário → decimal (MSB à esquerda)
Replica a lógica do R em Python: valida bits, imprime expansão legível e converte via acumulação esquerda→direita.

```python
from typing import Iterable, List

def bin_to_dec(bits: Iterable[int], msb_left: bool = True) -> int:
    bits_list: List[int] = list(bits)
    if len(bits_list) == 0:
        return 0
    if any(b not in (0, 1) for b in bits_list):
        raise ValueError("'bits' deve conter apenas 0 ou 1")
    if not msb_left:
        bits_list = list(reversed(bits_list))
    acc = 0
    for b in bits_list:
        acc = 2 * acc + int(b)
    return acc


def bin_expansion_string(bits: Iterable[int]) -> str:
    bits_list = list(bits)
    n = len(bits_list)
    if n == 0:
        return "0"
    exps = list(range(n - 1, -1, -1))
    terms = [f"{bits_list[i]}*2^{exps[i]}" for i in range(n) if bits_list[i] == 1]
    return " + ".join(terms) if terms else "0"
```

```python
# Exemplo de uso
bits = [1, 0, 1, 1, 1, 0, 0, 1]  # 10111001 (MSB à esquerda)
print("Bits:           ", ''.join(map(str, bits)))
print("Expansão:       ", bin_expansion_string(bits))
valor = bin_to_dec(bits)
print("Decimal (base10): ", valor)
```
