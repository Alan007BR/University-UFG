# Exercício 4 — Gauss–Jacobi e Gauss–Seidel
Vamos resolver o sistema do enunciado usando ambos os métodos e verificar a dominância diagonal. Critério: d = max |x_i^(k) − x_i^(k−1)| < 0,001; chute inicial x^(0) = (1,1,1,1); resultados com 6 casas.

```python
#a) Verificação de dominância diagonal (por linha)
import numpy as np
np.set_printoptions(suppress=True)

# Dados do exercício (A, b)
A = np.array([
    [ 5, -2,  0, -1],
    [ 1,  8,  2,  0],
    [ 0,  3,  7, -2],
    [ 0,  0,  3,  7]
], dtype=float)

b = np.array([2, -11, -27, 29], dtype=float)

# (a) Verificação de dominância diagonal (por linha)
row_sums = np.sum(np.abs(A), axis=1) - np.abs(np.diag(A))
alpha = row_sums / np.abs(np.diag(A))
print("alphas por linha (alpha1, alpha 2, etc..):", alpha)
print("max(alpha) =", np.max(alpha))
print("Estritamente dominante por linha? ", np.all(np.abs(np.diag(A)) > row_sums))
```

```python
#b) Método de Gauss–Jacobi
import numpy as np
np.set_printoptions(precision=6, suppress=True)

# ----------------------------------------------
# MÉTODO DE GAUSS–JACOBI
# ----------------------------------------------
# Critério de parada: d = max |x^(k) − x^(k−1)| < 0,001
# Chute inicial: x^(0) = (1,1,1,1)
# Sistema do enunciado (4x4)
# ----------------------------------------------

# Precisão desejada (erro máximo permitido)
e = 1e-3

# Chute inicial
x0 = np.array([[1.0],
               [1.0],
               [1.0],
               [1.0]])

# Matriz dos coeficientes A (4x4)
A = np.array([
    [ 5., -2.,  0., -1.],
    [ 1.,  8.,  2.,  0.],
    [ 0.,  3.,  7., -2.],
    [ 0.,  0.,  3.,  7.]
], dtype=float)

# Vetor de termos independentes B (4x1)
B = np.array([
    [  2.],
    [-11.],
    [-27.],
    [ 29.]
], dtype=float)

print("Matriz A:")
print(A)
print("\nVetor B:")
print(B)
print("\nChute inicial x0:")
print(x0)

print("\nInício do método de Gauss–Jacobi:\n")

# Número de variáveis (tamanho do sistema)
n = len(B)

# Vetores para armazenar os valores atuais e os anteriores
x_ant = x0.copy()
x_novo = np.zeros_like(x_ant)

# Contador de iterações
k = 0

# Loop principal — repete até a diferença entre iterações ser pequena
while True:
    k += 1  # conta a iteração atual
    
    # Calcula cada componente x_i
    for i in range(n):
        soma = 0
        
        # Soma dos termos A[i][j] * x_ant[j], exceto j = i
        for j in range(n):
            if j != i:
                soma += A[i][j] * x_ant[j]
        
        # Fórmula do método de Jacobi
        x_novo[i] = (B[i] - soma) / A[i][i]
    
    # Calcula o erro (diferença máxima entre iterações)
    erro = np.max(np.abs(x_novo - x_ant))
    
    # Mostra o progresso de cada iteração (6 casas)
    print(f"Iteração {k}: x = {x_novo.ravel()}, erro = {erro:.6f}")
    
    # Verifica o critério de parada
    if erro < e:
        break
    
    # Atualiza o vetor anterior
    x_ant = x_novo.copy()

# Exibe o resultado final
print("\nSolução aproximada encontrada:")
print(x_novo)
print(f"Número total de iterações: {k}")
```

```python
#c) Método de Gauss–Seidel
# Gauss–Seidel Method
import numpy as np
np.set_printoptions(precision=6, suppress=True)

# Precisão desejada (erro máximo permitido)
e = 1e-3

# Chute inicial (4x1)
x = np.array([[1.0],
              [1.0],
              [1.0],
              [1.0]])

# Matriz dos coeficientes A (4x4)
A = np.array([
    [ 5., -2.,  0., -1.],
    [ 1.,  8.,  2.,  0.],
    [ 0.,  3.,  7., -2.],
    [ 0.,  0.,  3.,  7.]
], dtype=float)

# Vetor de termos independentes B (4x1)
B = np.array([
    [  2.],
    [-11.],
    [-27.],
    [ 29.]
], dtype=float)

# Número de variáveis (tamanho do sistema)
n = len(B)

# Contador de iterações
k = 0

while True:
    k += 1  
    x_ant = x.copy()
    
    for i in range(n):
        soma = 0.0
        
        for j in range(n):
            if j != i:
                soma += A[i][j] * x[j]  # usa valores já atualizados quando j<i
        
        x[i] = (B[i] - soma) / A[i][i]
    
    erro = np.max(np.abs(x - x_ant))
    
    print(f"Iteração {k}: x = {x.ravel()}, erro = {erro:.6f}")
    
    # Verifica o critério de parada
    if erro < e:
        break
    
# Exibe o resultado final
print("\nSolução aproximada encontrada:")
print(x)
print(f"Número total de iterações: {k}")
```
