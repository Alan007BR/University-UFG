# Método da Eliminação de Gauss

O método de Gauss é uma técnica algébrica para resolver sistemas lineares do tipo $Ax = b$. Ele transforma o sistema original em um sistema equivalente, mas mais simples, usando operações elementares nas linhas da matriz.

## Teorema Fundamental

Dado um sistema $Ax = b$, podemos aplicar operações elementares nas equações para obter um novo sistema $\tilde{A}\tilde{x} = \tilde{b}$, equivalente ao original. As operações permitidas são:

1. Trocar duas equações;
2. Multiplicar uma equação por uma constante não nula;
3. Adicionar um múltiplo de uma equação a outra.

Essas operações não alteram o conjunto de soluções do sistema.

## Diferença entre métodos de resolução de sistemas lineares

Existem vários métodos para resolver sistemas lineares, como:

- **Métodos diretos**: Encontram a solução em um número finito de passos, como a eliminação de Gauss e a decomposição LU.
- **Métodos iterativos**: Aproximam a solução por meio de sucessivas aproximações, como Jacobi, Gauss-Seidel e métodos de gradiente.

O método de Gauss é um método direto, pois transforma a matriz $A$ em uma matriz triangular superior, permitindo resolver o sistema por retrossubstituição.

## Etapas do método de Gauss

1. **Triangularização**:
   - A matriz $A$ é transformada em uma matriz triangular superior usando operações elementares.
   - Em cada etapa $k$, elimina-se a variável $x_k$ das equações abaixo da linha $k$.
   - O elemento $a_{kk}$ é chamado de pivô da etapa.

2. **Retrossubstituição**:
   - Após triangularizar, resolve-se o sistema de cima para baixo, encontrando os valores das incógnitas.

A eliminação é feita por colunas, e os multiplicadores $m_{ik}$ são usados para zerar os elementos abaixo do pivô.

### Operação elementar: $l_2 \leftarrow l_2 - m_{21} l_1$

Esta operação é usada na eliminação de Gauss para zerar o elemento abaixo do pivô na coluna 1.
- $l_1$ e $l_2$ são linhas do sistema.
- $m_{21}$ é o multiplicador calculado: $m_{21} = \frac{a_{21}}{a_{11}}$.
- Subtraímos da linha 2 a linha 1 multiplicada por $m_{21}$, eliminando a variável $x_1$ da segunda equação.

**Exemplo:**
Se a matriz tem $a_{21}$ diferente de zero, após a operação, o novo $a_{21}$ será zero, facilitando a triangularização da matriz.

## Observações importantes sobre o método de Gauss

- O método exige que o pivô de cada etapa seja diferente de zero ($a_{kk} \neq 0$).
- Se algum pivô for zero, é necessário trocar linhas (pivoteamento) para garantir a continuidade do processo.
- O método de Gauss é eficiente para sistemas pequenos e médios, mas pode ser sensível a erros numéricos em sistemas grandes ou mal condicionados.

**Resumo:** O método de Gauss é um método direto, algébrico e sistemático para resolver sistemas lineares, transformando o sistema original em um sistema equivalente triangular, facilitando a obtenção da solução por retrossubstituição.

```python
# Algoritmo de Eliminação de Gauss (sem pivoteamento)
import numpy as np

def gauss_eliminacao(A, b):
    """
    Resolve o sistema linear Ax = b usando eliminação de Gauss sem pivoteamento.
    A: matriz quadrada (n x n)
    b: vetor coluna (n)
    Retorna vetor solução x.
    """
    A = A.astype(float).copy()  # Garante float e não altera original
    b = b.astype(float).copy()
    n = A.shape[0]
    # Etapa de eliminação
    for k in range(n-1):
        for i in range(k+1, n): # sl2 <- l2 - m21 * l1
            m = A[i, k] / A[k, k]  # Calcula multiplicador
            A[i, k] = 0            # Zera elemento abaixo da diagonal
            for j in range(k+1, n):
                A[i, j] = A[i, j] - m * A[k, j]  # Atualiza linha i
            b[i] = b[i] - m * b[k]               # Atualiza termo independente
    # Retrossubstituição
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        s = sum(A[i, j] * x[j] for j in range(i+1, n))
        x[i] = (b[i] - s) / A[i, i]
    return x

# Exemplo de uso:
A = np.array([[3, 2, 4],
              [0, 1/3, 2/3],
              [0, 0, 8]], dtype=float)
b = np.array([1, 5/3, 0], dtype=float)
x = gauss_eliminacao(A, b)
print("Solução do sistema:", x)
```

## Exercício: Eliminação de Gauss passo a passo (Sistema 4x4)

Objetivo: produzir as matrizes aumentadas \(A^{(0)}|b^{(0)}, A^{(1)}|b^{(1)}, \dots, A^{(3)}|b^{(3)}\) e resolver o sistema final triangular \(A^{(3)}x = b^{(3)}\).

Sistema original:
\n$\begin{cases}
2x_1 - 2x_2 - x_3 + 3x_4 = -7\\
x_1 - 2x_2 + 2x_3 + 2x_4 = -14\\
-5x_1 - 2x_2 - 8x_3 - 10x_4 = 8\\
-2x_1 + 3x_2 + 2x_3 - x_4 = -1
\end{cases}$

A seguir implementamos uma versão que guarda cada etapa.

```python
import numpy as np
np.set_printoptions(suppress=True)

def gauss_etapas(A, b, imprimir_m=True, imprimir_ab=True):
    A = A.astype(float).copy()
    b = b.astype(float).copy()
    n = A.shape[0]
    etapas = []  # lista de tuplas (A^{(k)}, b^{(k)})

    # A^{0} | b^{0}
    if imprimir_ab:
        print("A^{0} | b^{0} :")
        print("A^{0} :")
        print(A)
        print("b^{0} :")
        print(b)
    etapas.append((A.copy(), b.copy()))  # etapa 0

    # Etapas k = 0..n-2
    for k in range(n-1):
        piv = A[k,k]
        if abs(piv) < 1e-14:
            raise ZeroDivisionError(f"Pivô nulo na etapa {k}. Necessário pivoteamento.")

        # pivô + multiplicadores antes de imprimir A^{k+1}|b^{k+1}
        if imprimir_m:
            print("pivô")
            print(f"a[{k+1},{k+1}] = {piv:.6g}")
            print("multiplicadores")

        for i in range(k+1, n):
            m = A[i,k]/piv
            if imprimir_m:
                print(f"m[{i+1},{k+1}] = {m:.6g}")
            # aplica eliminação
            A[i,k] = 0.0
            for j in range(k+1, n):
                A[i,j] -= m*A[k,j]
            b[i] -= m*b[k]

        etapas.append((A.copy(), b.copy()))

        if imprimir_ab:
            print(f"A^{{{k+1}}} | b^{{{k+1}}} :")
            print(f"A^{{{k+1}}} :")
            print(A)
            print(f"b^{{{k+1}}} :")
            print(b)

    return etapas


def retrossubstituicao(U, y):
    n = U.shape[0]
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        s = 0.0
        for j in range(i+1, n):
            s += U[i,j]*x[j]
        x[i] = (y[i]-s)/U[i,i]
    return x

# Dados do sistema 4x4 (da imagem)
A0 = np.array([[  2, -2, -1,  3],
               [  1, -2,  2,  2],
               [ -5, -2, -8, -10],
               [ -2,  3,  2, -1]], dtype=float)
b0 = np.array([ -7, -14,  8, -1], dtype=float)

# Executa as etapas na ordem pedida
etapas = gauss_etapas(A0, b0, imprimir_m=True, imprimir_ab=True)

# Resolve com retrossubstituição usando a última etapa
A_final, b_final = etapas[-1]
x = retrossubstituicao(A_final, b_final)
print("Solução x:", x)
```
